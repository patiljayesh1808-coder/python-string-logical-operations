##### 1) Take a string and print it in reverse.

```python
string="Jayesh"

reverse = string[::-1]

print(reverse)
```

##### 2) Check whether a string is a palindrome.

```python
string = input("Enter a string:")

if string == string[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

##### 3) Count vowels and consonants in a string.

```python
string = input("Enter a String:")

vowels = 0
consonants = 0

for char in string:
    if char.lower() in "aeiouAEIOU":
        vowels += 1
    elif char.isalpha():
        consonants += 1

print("vowels in string:",vowels)
print("consonants in string:",consonants)
```

##### 4) Find how many times each character occurs.

```python
text = "programming"

for char in text:
    if text.count(char) > 0:
        print(char, ":" , text.count(char))
```

##### 5) Find characters that occur more than once.

```python
text = "programming"

for char in set(text):
    if text.count(char) > 1:
        print(char)
```

 ##### 6) Remove duplicate characters while keeping the first occurrence.

```python
string = "programming"

result = ""

for char in string:
    if char not in result:
        result += char

print(result)
```

##### 7) Find the longest word from a sentence.

```python
sentence = "python is very simple programming language"

words = sentence.split()

longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest Word:",longest)
```

##### 8) Check whether two strings contain the same characters regardless of order.

```python
str1 = "earth"          #silent - listen  ,  # care - race ,  # cat - act
str2 = "heart"

if sorted(str1) == sorted(str2):
    print("same characters")
else:
    print("Different characters")
```

##### 9) Completely remove characters that appear more tahn once from the given string.

```python
text = "programming"

result = ""

for char in text:
    if text.count(char) == 1:
        result += char

print(result)
```

##### 10) Reverse each word in a sentence individually,but do not change the position of the words. 

```python
text = "Python is easy language"

words = text.split()

result = ""

for word in words:
    result += word[::-1] + ""

print(result)
```
